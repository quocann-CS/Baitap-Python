### Problem 4
#### Write a function that returns the count of positive integers k such that 1 <= k < n and k > (n - 1) / 2. (Example: n=9 -> 4).

```python
def compute(n):
    count = 0
    for k in range(1, n):
        if k > (n - 1) / 2:
            count += 1
    return count
```

### Problem 6
#### Write a function that computes the area of a triangle with side lengths a, b, c (use Heron's formula). Raise ValueError if the sides cannot form a triangle.

```python
import math

def compute(a, b, c):
    # Kiểm tra điều kiện tạo thành tam giác
    if a <= 0 or b <= 0 or c <= 0 or (a + b <= c) or (a + c <= b) or (b + c <= a):
        raise ValueError("Độ dài các cạnh không thể tạo thành một tam giác.")

    # Tính nửa chu vi s và diện tích theo công thức Heron
    s = (a + b + c) / 2
    return math.sqrt(s * (s - a) * (s - b) * (s - c))
```

### Problem 8
#### Write a function that returns the character that is n positions after character x in the alphabet.
#### Preserve case and wrap around within A..Z or a..z.

```python
def compute(x, n):
    # Xử lý cho chữ cái viết thường
    if 'a' <= x <= 'z':
        return chr((ord(x) - ord('a') + n) % 26 + ord('a'))
    # Xử lý cho chữ cái viết hoa
    elif 'A' <= x <= 'Z':
        return chr((ord(x) - ord('A') + n) % 26 + ord('A'))
    return x
```

### Problem 9
#### Write a function that converts an uppercase letter to lowercase. If the input is not a single character, raise ValueError.

```python
def compute(s):
    # Kiểm tra nếu đầu vào không phải là đúng 1 ký tự
    if len(s) != 1:
        raise ValueError("Đầu vào phải là một ký tự đơn.")

    # Chuyển đổi thành chữ thường
    return s.lower()
```

### Problem 12
#### Write a function that returns a right isosceles triangle of height n using '*'.
#### Line i (1-indexed) has i '*', joined with newlines.
#### Example with n = 4
#### *
#### **
#### ***
#### ****

```python
def compute(n):
    # Tạo danh sách các dòng với i dấu '*' và nối lại bằng dấu xuống dòng '\n'
    return '\n'.join('*' * i for i in range(1, n + 1))
```
